When we replace missing value with some meaningful values this process is called imputation........when tha data contains numeric values then we can fill these missing space with mean median mode constant valuess this process is called NUMERIC IMPUTATION

```python
import pandas as pd
```

```python
df=pd.DataFrame({'Age':[10,20,30,80,50]})
```

```python
#MEAN IMPUTATION
#when we fill missing value with the mean
df['Age']:df['Age'].fillna(df['Age'].mean())
```

```python
#MEDIAN IMPUTATION
#when we fill missing value with the median
df['Age']:df['Age'].fillna(df['Age'].median())
```

```python
#MODE IMPUTATION
#when we fill missing value with the mode
df['Age']:df['Age'].fillna(df['Age'].mode()[0])
```

```python
#CONSTANT VALUE IMPUTATION
#when we fill missing value with the constant value
df['Age']:df['Age'].fillna(-1)
```

**1. Mean Imputation**

- Missing values ko average se replace karta hai.
- **Kab use karein?** Jab data normally distributed ho aur extreme outliers na hon.
- Code: `df['Age'].fillna(df['Age'].mean())`

**2. Median Imputation**

- Missing values ko middle value se replace karta hai.
- **Kab use karein?** Jab data mein outliers ho ya data skewed ho.
- Code: `df['Age'].fillna(df['Age'].median())`

**3. Mode Imputation**

- Missing values ko sabse zyada repeat hone wali value se replace karta hai.
- **Kab use karein?** Jab numerical data discrete ho ya koi value baar-baar repeat hoti ho.
- Code: `df['Age'].fillna(df['Age'].mode()[0])`

**4. Constant Value Imputation**

- Missing values ko kisi fixed value (jaise 0 ya -1) se replace karta hai.
- **Kab use karein?** Jab missing value kisi special condition ko represent karti ho ya use alag identify karna ho.
- Code: `df['Salary'].fillna(-1)`

### Easy Trick to Remember

- **Mean** = Average (Ausat)
- **Median** = Beech ki value
- **Mode** = Sabse zyada repeat hone wali value
- **Constant** = Fixed value

```python
#ENCODING
#MACHINE LEARNING ALGORITHM cannot understand the text categories so to make them understand we assign them numeric values to represent their categories
#to convert text into numbers is called encoding
```

```python
#label encoding : this assign unique number to every categories
df1=pd.DataFrame({'colours':["blue","red","green"]})
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df1['colours']=le.fit_transform(df1['colours'])
df1
```

```python
#ORDINAL ENCODING
#RANK WISE JESE schooling<graduation<post graduation=0<1<2 aise number assign hona
#Best For: Education Level, Ratings (Low/Med/High), Experience Level (Beginner/Expert).
```

```python
df2=pd.DataFrame({'education':['Graduate', 'PhD', 'Post Graduate', 'Graduate']})
```

```python
from sklearn.preprocessing import OrdinalEncoder
encoder = OrdinalEncoder(categories=[['Graduate', 'Post Graduate', 'PhD']])
df2[['education']] = encoder.fit_transform(df2[['education']])
print(df2)
```

```python
#One-Hot Encoding
#Creates a new binary column for every unique category. The original column is replaced by a matrix of 1s and 0s. Most widely used method.
```

```python
import numpy as np

X = np.array([[10], [20], [30], [40], [50],[500]])
```

standard scaler is that when we ignore the largest value in a dataset as it doesnt match the other datasets value and ignore that value so that it doesnt create unnecessary influence on mean

```python
from sklearn.preprocessing import StandardScaler
scaler= StandardScaler()
X_scaled = scaler.fit_transform(X)
print(X_scaled)
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler=MinMaxScaler()
X_scaled=scaler.fit_transform(X)
print(X_scaled)
```

```python
from sklearn.preprocessing import RobustScaler

scaler = RobustScaler()
X_scaled = scaler.fit_transform(X)
print(X_scaled)
# Uses median & IQR — outlier resistant!
```

```python
from sklearn.preprocessing import MaxAbsScaler

scaler = MaxAbsScaler()
X_scaled = scaler.fit_transform(X)
print(X_scaled)
# Scales to [-1, 1], preserves zero entries
```

```python
from sklearn.preprocessing import Normalizer

scaler = Normalizer(norm='l2')
X_scaled = scaler.fit_transform(X)
print(X_scaled)

# Each row becomes a unit-length vector
```

Outlier Handling
An outlier is a data point that lies far from the rest of the data. A single outlier can shift the mean, corrupt scaling, and skew every model trained on it.

Why Are Outliers Dangerous?
Shifts the Mean
Salary [30,40,50,60] → mean = 45
Add 5000 → mean = 1036
One point destroys the average.
Distorts Scaling
MinMaxScaler maps values to [0,1].
With outlier 5000, all normal values
get compressed to near 0.00–0.01.
Skews Regression
The regression line pulls toward
the outlier. Predictions on normal
data become inaccurate.
Inflates Variance
One extreme value explodes std dev,
making data appear highly variable
even when most values are tight.
